### **Preparación de los datos**

```python
# Importación de librerías
import pandas as pd # Trabajar con datos estructurados
import numpy as np
from IPython.display import display, HTML # Visualización de objetos
import matplotlib.pyplot as plt # Visualización de gráficos
import seaborn as sns # Visualización de gráficos
import warnings
warnings.filterwarnings("ignore", category=FutureWarning)
```

```python
# Lectura de la data
data = pd.read_csv('cancer.csv')
data.head()
```

```python
# Eliminacion de primera columna
col = data.columns[0]
data = data.drop(col, axis=1)
data.head()
```

```python
# Se revisan los valores faltantes
data.isnull().sum()
```

```python
# Se corrobora el tipo de datos
data.info()
```

```python
# Se revisan los valores únicos de la variable objetivo 'Class'
data['Class'].unique()
```

```python
# Se convierte la variable objetivo ('Class') a categórica
data['Class'] = data['Class'].astype('category')
```

```python
# Se separan las características (X) de la variable objetivo (y)
X = data.select_dtypes(include='number')
y = data['Class']
```

```python
# Se escalan las características
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Se dividen los datos en conjuntos de entrenamiento y prueba
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.25, random_state=42)
```

### **Modelo de red neuronal**

```python
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score

# Ajuste de hiperparámetros
# Se van a evaluar las siguientes arquitecturas de capas ocultas en la red neuronal:
# 1) 1 capa oculta de 5 neuronas
# 2) 1 capa oculta de 10 neuronas
# 3) 2 capas ocultas de 5 neuronas cada una
# 4) primera capa oculta de 10 neuronas y segunda oculta de 5 neuronas
# 5) primera capa oculta de 5 neuronas y segunda oculta de 10 neuronas
# 6) 2 capas ocultas de 10 neuronas cada una

# Se evaluan 3 cantidades de iteraciones: 200, 300 y 400

# Se prueban 2 funciones de activación:
# 1) Relu: f(x)=max(0,x)
# 2) Tangente hiperbólica: f(x)=tanh(x)

best_accuracy = 0.0 
best_model = None 
hidden_layer_sizes_options = [(5,), (10,), (5, 5), (10, 5), (5, 10), (10, 10)]
max_iter_options = [200, 300, 400]
activation_options = ['relu', 'tanh']
solver_options = ['adam', 'lbfgs'] # Se manejan dos tipos de solver para la actualización de los pesos sinápticos y bias en cada iteración

print("Comenzando la búsqueda de los mejores hiperparámetros...")

for hidden_layers in hidden_layer_sizes_options:
    for max_iter in max_iter_options:
        for activation in activation_options:
            for solver in solver_options:
                mlp = MLPClassifier(hidden_layer_sizes=hidden_layers, max_iter=max_iter,
                                    activation=activation, solver=solver, random_state=42)
                mlp.fit(X_train, y_train)
                y_pred = mlp.predict(X_test)
                accuracy = accuracy_score(y_test, y_pred)
                print(f"Capa(s) oculta(s): {hidden_layers}, Iteraciones: {max_iter}, Activación: {activation}, Solver: {solver}, Precisión: {accuracy:.4f}")
                if accuracy > best_accuracy:
                    best_accuracy = accuracy
                    best_model = mlp
                    best_params = {'hidden_layers': hidden_layers, 'max_iter': max_iter,
                                   'activation': activation, 'solver': solver}

# 5. Evaluar el mejor modelo
print("\nMejor modelo encontrado:")
print(f"Parámetros: {best_params}")
print(f"Precisión en el conjunto de prueba del mejor modelo: {best_accuracy:.4f}")
```

El proceso de búsqueda de hiperparámetros encontró que una red neuronal con una capa oculta de 10 neuronas, utilizando la función de activación ReLU y el optimizador Adam, entrenada durante un máximo de 300 iteraciones, logró una precisión de aproximadamente el 97.14% al clasificar muestras no vistas de cáncer de mama. Este desempeño es muy bueno para el tema de predicción de células cancerígenas.
